```python
def clean_currency(series):
    return pd.to_numeric(
        series.astype("string")
              .str.replace(
                  "₱", "", regex=False)
              .str.replace(
                  ",", "", regex=False)
              .str.strip(),
        errors="coerce"
    )
```

```python
#6 #Build Analysis-Ready Fields from Cleaned Tables

items_analysis = items_clean.copy()

items_analysis["purchase_price_num"] = (
    clean_currency(
        items_analysis["unit_price_at_purchase"]
    )
)

items_analysis = items_analysis[
    (items_analysis["quantity"] > 0)
    & items_analysis["purchase_price_num"].notna()
]

items_analysis = items_analysis.merge(
    products_clean[["product_id", "unit_cost", "category_clean"]],
    on="product_id",
    how="left"
)

items_analysis = items_analysis.merge(
    orders_clean[["order_id", "sales_channel"]],
    on="order_id",
    how="left"
)

items_analysis["line_revenue"] = (
    items_analysis["quantity"]
    * items_analysis["purchase_price_num"]
)

items_analysis["line_cost"] = (
    items_analysis["quantity"]
    * items_analysis["unit_cost"]
)

items_analysis["line_profit"] = (
    items_analysis["line_revenue"]
    - items_analysis["line_cost"]
)
```

```python
items_analysis
```

```python
#Compare Sales Channels with groupby()

sales_analysis = items_analysis.merge(
    orders_clean[["order_id", "order_status"]],
    on="order_id",
    how="left"
)

channel_summary = (
    sales_analysis
    .groupby("sales_channel")
    .agg(
        gross_line_revenue=("line_revenue", "sum"),
        average_line_revenue=("line_revenue", "mean"),
        line_items=("order_item_id", "count"),
        orders=("order_id", "nunique")
    )
    .sort_values(
        "gross_line_revenue", ascending=False
    )
)

print(channel_summary)
```

```python
#EDA 1 - Which sales channel contributes most gross item revenue?

channel_summary = (
    items_analysis
    .groupby("sales_channel")
    .agg(
        gross_line_revenue=("line_revenue", "sum"),
        orders=("order_id", "nunique")
    )
    .sort_values("gross_line_revenue", ascending=False)
)

print(channel_summary)
```

EDA#1 findings: The Web sales channel contributes the most gross item revenue, totaling 2,324,877.34 across 4,885 unique orders. However, gross revenue alone does not indicate which channel is the most profitable.

```python
#EDA 2 - Which cleaned product category has highest item revenue?


category_summary = (
    items_analysis
    .groupby("category_clean")
    .agg(
        gross_line_revenue=("line_revenue", "sum"),
        line_items=("order_item_id", "count")
    )
    .sort_values(
        "gross_line_revenue", ascending=False
    )
)

print(category_summary)
```

EDA#2 findings: Toys generated the highest gross line revenue among the product categories, exceeding Electronics by ₱23,428.91. Electronics had the most line items, with 5,045, showing that the category with the most line items is not necessarily the one with the highest revenue.
